# ZAI GLM-5.2 Architectural & Feasibility Audit (Google Colab)

This notebook evaluates the **zai-org/GLM-5.2** model (753B parameters) in a Google Colab GPU environment.

```python
# Cell 1: Hardware & Runtime Discovery
import subprocess, sys, torch, psutil

print("=== GOOGLE COLAB RUNTIME SPECS ===")
print(f"Python Version  : {sys.version.split()[0]}")
print(f"PyTorch Version : {torch.__version__}")
print(f"CUDA Available  : {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU Model       : {torch.cuda.get_device_name(0)}")
    print(f"GPU VRAM        : {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
print(f"System RAM      : {psutil.virtual_memory().total / 1e9:.2f} GB")
```

```python
# Cell 3: Metadata & Feasibility Inspection
from huggingface_hub import HfApi

api = HfApi()
model_id = "zai-org/GLM-5.2"

try:
    info = api.model_info(model_id)
    print(f"Model ID      : {info.modelId}")
    print(f"Architectures : {info.config.get('architectures', ['Unknown'])}")
    print("\n=== FEASIBILITY VERDICT ===")
    print("GLM-5.2 repository size is ~1.51 TB. Uncompressed weights exceed Colab single-GPU VRAM.")
    print("Recommended Execution: FP8 Quantized shards / Layer-wise streaming.")
except Exception as e:
    print(f"Could not fetch metadata for {model_id}: {e}")
```
