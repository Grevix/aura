# AURA Experiment Lab: 02 — Local Model Inspection

Local consumer target hardware: Intel Core i5-13420H (8/12 cores), 16.79 GB RAM, NVMe SSD @ 2.3 GB/s.

```python
import subprocess

models = ['qwen2.5-coder:7b', 'llama3:8b-instruct-q4_0', 'deepseek-r1:latest', 'mistral:latest', 'gemma4:latest']
for m in models:
    print(f'--- Inspecting Local Model: {m} ---')
    res = subprocess.run(['cargo', 'run', '-p', 'aura-cli', '--', 'plan', '-m', m, '-b', '4G'], capture_output=True, text=True)
    print(res.stdout)
```
